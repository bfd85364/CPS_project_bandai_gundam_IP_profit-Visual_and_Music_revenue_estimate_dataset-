```python
import numpy as np
import pandas as pd
from pandas import DataFrame as df
```

```python
data_path = '/content/drive/MyDrive/CPS_GUNDAM_DATA/'
raw_train_FullYear_group = pd.read_csv(data_path + 'Full_year_groupwide_gundam_sales_train.csv') #훈련용 연도별 전체 그룹부문 IP 매출액 데이터
raw_test_FullYear_group = pd.read_csv(data_path + 'Full_year_groupwide_gundam_sales_test.csv') #테스트용 연도별 전체 그룹부문 IP 매출액 데이터

#데이터 전처리 1단계-결측값(NaN) 제거
train_FullYear_group = raw_train_FullYear_group.fillna(0) #훈련용 연도별 전체 그룹부문 IP 매출액 데이터
test_FullYear_group = raw_test_FullYear_group.fillna(0) #테스트용 연도별 전체 그룹부문 IP 매출액 데이터

print("훈련데이터")
display(train_FullYear_group)
print("=====================================================================\n")
print("테스트데이터")
display(test_FullYear_group)
```

```python
#연도별 IP별 완구 부문 매출액
raw_train_FullYear_toys = pd.read_csv(data_path + 'Full_year_Toys_japan_gundam_sales_train.csv') #학습(훈련)용 전체년도 IP별 완구부문 매출액 데이터
raw_test_FullYear_toys = pd.read_csv(data_path + 'Full_year_Toys_japan_gundam_sales_test.csv') #테스트용 전체년도 IP별 완구부문 매출액 데이터

#데이터 전처리 1단계-결측값(NaN) 제거
train_FullYear_toys= raw_train_FullYear_toys.fillna(0)
test_FullYear_toys= raw_test_FullYear_toys.fillna(0)

print("훈련데이터")
display(train_FullYear_toys)
print("=====================================================================\n")
print("테스트데이터")
display(test_FullYear_toys)
```

```python
#r복원한 건담 게임 (콘솔, 모바일, pc게임 종합) 연간 매출 추정치
raw_game_years_data = pd.read_csv(data_path + 'gundam_game_revenue_dataset.csv')
raw_game_years_data.head()
```

```python
#데이터 전처리: 훈련 데이터,테스트 데이터 분할

print("훈련 데이터")
train_game_years_data = raw_game_years_data[(raw_game_years_data['연도'] <= 2026) & (raw_game_years_data['연도'] > 2007)] #영상, 음악 추정치 데이터를 구하기 위해서 전체 그룹 부문에서 존재 하지 않는 2006년, 2007년 데이터는 제외
display(train_game_years_data)

print("=====================================================================\n")

print("테스트 데이터")
test_game_years_data = raw_game_years_data[raw_game_years_data['연도'] > 2020]
display(test_game_years_data)
```

```python
#Visual_and_Music Content Business 항목의 매출 데이터 구하기
#안분(Allocation) 산출 공식 대신, 전체 그룹부문 매출액 - (완구부문 매출액 - 디지털 게임 부문 매출액 추정치) 공식으로 단순화

# 1. 단위를 맞추기 위해 게임 매출액(원래 엔화 단위)을 1억 엔(10^8) 단위로 변환
train_game_scaled = train_game_years_data.copy()
train_game_scaled['게임_매출(억엔)'] = train_game_scaled['추정_매출액(엔화)'] / 1e8

test_game_scaled = test_game_years_data.copy()
test_game_scaled['게임_매출(억엔)'] = test_game_scaled['추정_매출액(엔화)'] / 1e8

# 2. 훈련 데이터 병합 및 계산 (2008년 ~ 2020년 데이터 필터링 혹은 매칭)
train_merged = pd.merge(train_FullYear_group[['Full_Year', 'Mobile_Suit_Gundam']],
                       train_FullYear_toys[['Full_Year', 'Mobile_Suit_Gundam']],
                       on='Full_Year', suffixes=('_group', '_toys'))
train_merged = pd.merge(train_merged, train_game_scaled[['연도', '게임_매출(억엔)']],
                       left_on='Full_Year', right_on='연도')

# 공식 적용: 그룹 - (완구 - 게임)
train_merged['Visual_Music_Revenue'] = train_merged['Mobile_Suit_Gundam_group'] - (train_merged['Mobile_Suit_Gundam_toys'] - train_merged['게임_매출(억엔)'])

# 3. 테스트 데이터 병합 및 계산
test_merged = pd.merge(test_FullYear_group[['Full_Year', 'Mobile_Suit_Gundam']],
                      test_FullYear_toys[['Full_Year', 'Mobile_Suit_Gundam']],
                      on='Full_Year', suffixes=('_group', '_toys'))
test_merged = pd.merge(test_merged, test_game_scaled[['연도', '게임_매출(억엔)']],
                      left_on='Full_Year', right_on='연도')

# 공식 적용: 그룹 - (완구 - 게임)
test_merged['Visual_Music_Revenue'] = test_merged['Mobile_Suit_Gundam_group'] - (test_merged['Mobile_Suit_Gundam_toys'] - test_merged['게임_매출(억엔)'])

print("훈련용 영상 및 음악 라이선스 매출액 추정치:")
display(train_merged[['Full_Year', 'Visual_Music_Revenue']])

print("=====================================================================\n")

print("테스트용 영상 및 음악 라이선스 매출액 추정치:")
display(test_merged[['Full_Year', 'Visual_Music_Revenue']])

# 4. CSV 파일로 구글 드라이브에 내보내기
train_merged[['Full_Year', 'Visual_Music_Revenue']].to_csv(data_path + 'train_visual_music_revenue_estimated.csv', index=False, encoding='utf-8-sig')
test_merged[['Full_Year', 'Visual_Music_Revenue']].to_csv(data_path + 'test_visual_music_revenue_estimated.csv', index=False, encoding='utf-8-sig')

print("=====================================================================\n")
print("CSV 파일 내보내기 완료!")
```
